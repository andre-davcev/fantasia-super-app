```python
def calculate_sum(grades):
    sum_of_grades = round(sum(grades), 0)
    return int(sum_of_grades)
```

```python
def calculate_average(grades):
    average=calculate_sum(grades)/len(grades)
    return int(average)
```

```python
grades = [85, 70, 60, 90, 93.4,100]

returned_sum = calculate_sum(grades)
returned_average= calculate_average(grades)


print(returned_sum)
print(returned_average)

```
