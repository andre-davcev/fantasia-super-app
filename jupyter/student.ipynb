```python
class Student:
    # Constructor with first name and last name
    def __init__(self, first_name, last_name):
        self.set_first_name(first_name)
        self.set_last_name(last_name)
        self.grades = [] #Set grades to empty array

    # Method to set the first name
    def set_first_name(self, first_name):
        self.first_name = first_name.capitalize()
    
    # Method to get the first name
    def get_first_name(self):
        return self.first_name
    
    # Method to set the last name
    def set_last_name(self, last_name):
        self.last_name = last_name.capitalize()

    # Method to get the last name
    def get_last_name(self):
        return self.last_name
    
    # Method to print the last and first name separated by comma
    def print_full_name(self):
        print(f"{self.last_name}, {self.first_name}")

    # Method to add grade
    def add_grade(self, grade):
        return self.grades.append(grade)

    # Method to print all the students grades
    def print_grades(self):
        print("Grades")

        for i in range(len(self.grades)):
            print(f"  {i}: {self.grades[i]}")

    # Method to print grade average
    def print_average(self):
        grade_sum = sum(self.grades)
        grade_total = len(self.grades)
        average = grade_sum / grade_total
        average_rounded = round(average, 1)

        print(f"Grade Average: {average_rounded}")
```

```python
jane_doe = Student("jane", "doe")

# Print their full name
jane_doe.print_full_name()

# They got married, update their last name
jane_doe.set_last_name("smith")

# Re-Print their updated last name
jane_doe.print_full_name()
```

```python
# Add some grades
jane_doe.add_grade(87.4)
jane_doe.add_grade(63.5)
jane_doe.add_grade(99.0)
jane_doe.add_grade(100)
jane_doe.add_grade(80)

# Print your grades
jane_doe.print_grades()

# Print the average
jane_doe.print_average()
```
